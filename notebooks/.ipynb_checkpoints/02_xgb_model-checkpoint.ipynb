```python
import pandas as pd
import numpy as np
from xgboost import XGBClassifier
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import accuracy_score, log_loss, classification_report
import joblib
import os

# 1. Load the multi-season processed dataset
df = pd.read_csv('../data/processed/epl_multi_season_features.csv')
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values('Date').reset_index(drop=True)

# 2. Define Target (0 = Away Win, 1 = Draw, 2 = Home Win)
conditions = [
    df['FTHG'] > df['FTAG'],
    df['FTHG'] == df['FTAG'],
    df['FTHG'] < df['FTAG']
]
df['Target'] = np.select(conditions, [2, 1, 0], default=1)

# 3. Engineer Match-Context Features
# Highlighting the gap between attack and defense
df['xG_Attack_Diff'] = df['Home_xG_Created_ewma'] - df['Away_xG_Conceded_ewma']
df['xG_Defense_Diff'] = df['Away_xG_Created_ewma'] - df['Home_xG_Conceded_ewma']

# Expected Match xG (Low total = higher probability of a Draw)
df['Expected_Match_xG'] = df['Home_xG_Created_ewma'] + df['Away_xG_Created_ewma']

# Luck Differential (Who is currently overperforming their metrics more?)
df['Luck_Diff'] = df['Home_Luck_Factor_ewma'] - df['Away_Luck_Factor_ewma']

# Convert Bookmaker Odds into Normalized Implied Probabilities
raw_margin = (1 / df['B365H']) + (1 / df['B365D']) + (1 / df['B365A'])
df['Bookie_Prob_A'] = (1 / df['B365A']) / raw_margin
df['Bookie_Prob_D'] = (1 / df['B365D']) / raw_margin
df['Bookie_Prob_H'] = (1 / df['B365H']) / raw_margin

# Define final feature list
features = [
    'Home_xG_Created_ewma', 'Home_xG_Conceded_ewma',
    'Away_xG_Created_ewma', 'Away_xG_Conceded_ewma',
    'Home_Luck_Factor_ewma', 'Away_Luck_Factor_ewma',
    'xG_Attack_Diff', 'xG_Defense_Diff', 'Expected_Match_xG', 'Luck_Diff',
    'Elo_Diff', 'Rest_Diff', 
    'Bookie_Prob_H', 'Bookie_Prob_D', 'Bookie_Prob_A'
]

X = df[features]
y = df['Target']
```

```python
train_idx = int(len(df) * 0.70)
val_idx = int(len(df) * 0.80)

X_train, y_train = X.iloc[:train_idx], y.iloc[:train_idx]
X_val, y_val = X.iloc[train_idx:val_idx], y.iloc[train_idx:val_idx]
X_test, y_test = X.iloc[val_idx:], y.iloc[val_idx:]

print(f"Training: {len(X_train)} | Validation: {len(X_val)} | Testing: {len(X_test)}")
```

```python
# Base XGBoost model with stronger regularization (lower depth, higher L1/L2 penalties)
base_model = XGBClassifier(
    n_estimators=150,
    max_depth=2,          
    learning_rate=0.015,  
    reg_alpha=1.0,        
    reg_lambda=2.0,       
    subsample=0.7,
    colsample_bytree=0.7,
    objective='multi:softprob',
    num_class=3,
    random_state=42
)

base_model.fit(X_train, y_train)
print("XGBoost model training complete!")
```

```python
# 2. Use CalibratedClassifierCV with built-in cross-validation (cv=3)
# This trains and calibrates cleanly without needing a manual pre-fit split
print("Training and calibrating model with cross-validation...")
calibrated_model = CalibratedClassifierCV(
    estimator=base_model,
    method='isotonic',
    cv=3
)

# Fit directly on your combined training + validation dataset (or just X_train, y_train)
# Using the 80% combined training data block ensures it has plenty of samples for isotonic regression
train_val_idx = int(len(df) * 0.80)
X_tr = X.iloc[:train_val_idx]
y_tr = y.iloc[:train_val_idx]
X_test = X.iloc[train_val_idx:]
y_test = y.iloc[train_val_idx:]

calibrated_model.fit(X_tr, y_tr)
print("Model training & calibration complete!")
```

```python
# Generate predictions on the unseen Test set
y_probs = calibrated_model.predict_proba(X_test)
y_preds = calibrated_model.predict(X_test)

# Calculate log-loss and accuracy
model_loss = log_loss(y_test, y_probs)
model_acc = accuracy_score(y_test, y_preds)

# Calculate baseline Bookmaker metrics
bookie_probs = X_test[['Bookie_Prob_A', 'Bookie_Prob_D', 'Bookie_Prob_H']].values
bookie_loss = log_loss(y_test, bookie_probs)
bookie_acc = accuracy_score(y_test, np.argmax(bookie_probs, axis=1))

print("\n=== FINAL EVALUATION RESULTS ===")
print(f"Model Accuracy:    {model_acc:.1%}  |  Model Log-Loss:    {model_loss:.4f}")
print(f"Bookie Accuracy:   {bookie_acc:.1%}  |  Bookie Log-Loss:   {bookie_loss:.4f}\n")

print(classification_report(y_test, y_preds, target_names=['Away Win', 'Draw', 'Home Win']))

# Save the calibrated model
os.makedirs('../models', exist_ok=True)
joblib.dump(calibrated_model, '../models/calibrated_xgb_outcome.pkl')
print("Model saved to ../models/calibrated_xgb_outcome.pkl")
```

```python
import numpy as np
from sklearn.metrics import accuracy_score, classification_report

# We use the y_probs (probabilities) already generated by your calibrated model
# y_probs structure: [Prob_Away, Prob_Draw, Prob_Home]

def custom_match_predictor(probs, draw_threshold=0.28):
    """
    Forces the model to pick a Draw if the probability exceeds the threshold,
    otherwise defaults to the highest remaining probability.
    """
    custom_preds = []
    for row in probs:
        p_away, p_draw, p_home = row[0], row[1], row[2]
        
        # If the draw probability crosses our minimum threshold, pick Draw (1)
        if p_draw >= draw_threshold:
            custom_preds.append(1)
        # Otherwise, pick whoever is more likely between Home (2) and Away (0)
        elif p_home > p_away:
            custom_preds.append(2)
        else:
            custom_preds.append(0)
            
    return np.array(custom_preds)

# Apply the custom threshold
# (You can tweak 0.28 up or down to balance how many draws it picks)
y_preds_custom = custom_match_predictor(y_probs, draw_threshold=0.29)

# Recalculate Accuracy and Print Report
custom_acc = accuracy_score(y_test, y_preds_custom)

print("\n=== CUSTOM THRESHOLD RESULTS ===")
print(f"Original Accuracy: {model_acc:.1%} | Custom Accuracy: {custom_acc:.1%}")
print(f"Log-Loss remains unchanged at: {model_loss:.4f}\n")

# The zero_division=0 silences any lingering warnings
print(classification_report(y_test, y_preds_custom, target_names=['Away Win', 'Draw', 'Home Win'], zero_division=0))

# See how many of each it predicted this time
unique, counts = np.unique(y_preds_custom, return_counts=True)
print("\nNew Predictions Breakdown (0=Away, 1=Draw, 2=Home):")
print(dict(zip(unique, counts)))
```

```python
import numpy as np
from scipy.stats import poisson

def predict_exact_score(home_xg, away_xg, rho=-0.15, max_goals=5):
    """
    Calculates the probability matrix of exact scorelines using the Dixon-Coles 
    adjustment to standard Poisson distributions.
    """
    # 1. Create independent Poisson probabilities for each team (0 to 5 goals)
    home_probs = poisson.pmf(np.arange(max_goals + 1), home_xg)
    away_probs = poisson.pmf(np.arange(max_goals + 1), away_xg)
    
    # 2. Calculate the matrix of independent joint probabilities (Home x Away)
    score_matrix = np.outer(home_probs, away_probs)
    
    # 3. Apply the Dixon-Coles adjustment for low-scoring dependent matches
    score_matrix[0, 0] *= (1 - home_xg * away_xg * rho)
    score_matrix[1, 0] *= (1 + home_xg * rho)
    score_matrix[0, 1] *= (1 + away_xg * rho)
    score_matrix[1, 1] *= (1 - rho)
    
    # 4. Normalize the matrix so probabilities sum exactly to 1.0 after adjustment
    score_matrix /= np.sum(score_matrix)
    
    # 5. Extract the most likely exact scoreline (highest probability in the grid)
    max_prob_idx = np.argmax(score_matrix)
    pred_home_goals, pred_away_goals = np.unravel_index(max_prob_idx, score_matrix.shape)
    
    return pred_home_goals, pred_away_goals, score_matrix[pred_home_goals, pred_away_goals]
```

```python
# Assuming your X_test dataframe contains the EWMA xG features
predictions = []

for index, row in X_test.iterrows():
    # Use your previously engineered expected goals averages
    home_xg = row['Home_xG_Created_ewma']
    away_xg = row['Away_xG_Created_ewma']
    
    # In rare cases of edge data, ensure xG doesn't drop below a tiny minimum
    home_xg = max(home_xg, 0.1)
    away_xg = max(away_xg, 0.1)
    
    home_goals, away_goals, exact_prob = predict_exact_score(home_xg, away_xg)
    
    predictions.append({
        'Pred_Home_Goals': home_goals,
        'Pred_Away_Goals': away_goals,
        'Score_Probability': exact_prob,
        'Formatted_Score': f"{home_goals}-{away_goals}"
    })

scoreline_df = pd.DataFrame(predictions, index=X_test.index)

# Merge back with your original test set to review
results_df = X_test.copy()
results_df['Predicted_Score'] = scoreline_df['Formatted_Score']
results_df['Score_Confidence'] = scoreline_df['Score_Probability']

print(results_df[['Home_xG_Created_ewma', 'Away_xG_Created_ewma', 'Predicted_Score', 'Score_Confidence']].head(10))
```
